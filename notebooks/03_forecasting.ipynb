```python
import pandas as pd
import matplotlib.pyplot as plt
from src.forecasting.baselines import MovingAverageForecaster
from src.forecasting.statistical import HoltWintersForecaster

df = pd.read_parquet('data/processed/demand_daily.parquet')
final_fc = pd.read_csv('outputs/forecasts/final_forecasts.csv')
final_fc.head(10)
```

```python
# Visualize sample SKU forecast
sample = final_fc[(final_fc['sku_id'] == 'SKU-ELEC-001') & (final_fc['store_id'] == 'FC-East')]
plt.figure(figsize=(10, 4))
plt.plot(pd.to_datetime(sample['forecast_date']), sample['forecast'], label='Forecast', color='orange')
plt.fill_between(pd.to_datetime(sample['forecast_date']), sample['lower_bound'], sample['upper_bound'], color='orange', alpha=0.3, label='95% PI')
plt.title('SKU-ELEC-001 Forecast Trajectory (FC-East)')
plt.legend()
plt.show()
```
